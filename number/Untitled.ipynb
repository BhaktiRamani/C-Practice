```python
import random
x=random.randint(0,11)


while True :
    user_guess=input("make a guess ")
    print(user_guess)
    if user_guess.isdigit():
        y=int(user_guess)
        if y==x:
            print("ur guess is right")
    
        else:
            print("better luck next time")
            continue
            
    else:
        print("enter number next time")
        continue
```
